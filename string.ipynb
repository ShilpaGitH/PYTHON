##### 1. Reverse a string without using slicing.

```python
text = "Python"
# Expected output: nohtyP
reverse = ""

for i in text:
    reverse = i + reverse

print(reverse)
```

```python
text = "CAT"
reverse = ""

for i in text:
    reverse = reverse + i

print(reverse)
```

### You want to reverse the order of words, not reverse the letters. "This is a Cat"

```python
sentence = "This is a Cat"
sentence
```

```python
word = sentence.split()
```

```python
word
```

```python
new_word = word[::-1]
```

```python
new_word
```

```python
result = " ".join(new_word) 

### " " → A space between each word.
### .join() → Combines all elements of a list into one string.
```

```python
result
```
