### 1. Positive, Negative, or Zero
Write a program that accepts an integer from the user and determines whether it is positive,
negative, or zero.

```python
num = int(input("Enter the number:"))

if num > -1 and num < 1:
    print('zero')

elif num < 1:
    print('Negative')

else:
    print('positive')


    
```

### 2. Even or Odd
Write a program that checks whether a given integer is even or odd

```python
num = int(input("Enter the number:"))

if num%2 == 0:
    print(num,'even')
else:
    print(num, 'odd')
```

### 3. Greatest of Three Numbers
Accept three numbers from the user and print the largest among them using conditional statements

```python
num1 = int(input("Enter the number:"))
num2 = int(input("Enter the number:"))
num3 = int(input("Enter the number:"))

if num1>num2 and num1>num3:
    print(num1,"largest")
elif num2>num1 and num2>num3:
    print(num2,'largest')
else:
    print(num3,'largest')
```

### 4. Leap Year Checker
Write a program to determine whether a given year is a leap year according to the Gregorian
calendar rules

year 400 divisible - leap year
year 100 divisible - not leap year
year 4 divisible - leap year

```python
year = int(input("Enter the number:"))

if year % 400 == 0:
    print('leap year')
elif year % 100==0:
    print("not leap year")
elif year % 4 == 0:
    print('leap year')
else:
    print('Not leap year')

```
